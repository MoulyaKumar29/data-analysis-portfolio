```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
stock_data = {
    "Date": pd.date_range(start="2026-01-01", periods=15, freq="D"),
    "Open": [
        100, 102, 101, 105, 107, 106, 110, 108,
        112, 115, 114, 118, 116, 120, 122
    ],
    "High": [
        105, 104, 106, 108, 110, 111, 113, 112,
        115, 118, 117, 121, 120, 123, 126
    ],
    "Low": [
        98, 100, 99, 103, 104, 103, 107, 106,
        110, 112, 111, 115, 113, 118, 120
    ],
    "Close": [
        103, 101, 105, 107, 106, 110, 108, 112,
        114, 116, 115, 119, 118, 122, 125
    ],
    "Volume": [
        10000, 12000, 11000, 15000, 13000,
        18000, 16000, 20000, 22000, 25000,
        21000, 28000, 24000, 30000, 35000
    ]
}

stock_df = pd.DataFrame(stock_data)

stock_df.head()
```

```python
print("Dataset Shape:", stock_df.shape)

print("\nColumn Names:")
print(stock_df.columns)

print("\nDataset Information:")
stock_df.info()

print("\nStatistical Summary:")
stock_df.describe()
```

```python
stock_df["Daily_Return"] = stock_df["Close"].pct_change() * 100

print("Stock Prices and Daily Returns:")
print(stock_df[["Date", "Close", "Daily_Return"]].round(2))
```

```python
print("STOCK MARKET SUMMARY")
print("-" * 30)

print("Average Closing Price:", round(stock_df["Close"].mean(), 2))
print("Highest Closing Price:", stock_df["Close"].max())
print("Lowest Closing Price:", stock_df["Close"].min())
print("Average Daily Return (%):", round(stock_df["Daily_Return"].mean(), 2))
print("Highest Trading Volume:", stock_df["Volume"].max())
print("Lowest Trading Volume:", stock_df["Volume"].min())
print("Total Trading Volume:", stock_df["Volume"].sum())
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    stock_df["Date"],
    stock_df["Close"],
    marker="o"
)

plt.title("Stock Closing Price Trend")
plt.xlabel("Date")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    stock_df["Date"],
    stock_df["Daily_Return"]
)

plt.title("Daily Stock Returns (%)")
plt.xlabel("Date")
plt.ylabel("Daily Return (%)")
plt.xticks(rotation=45)
plt.axhline(y=0, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
daily_volatility = stock_df["Daily_Return"].std()

print("Daily Return Volatility:", round(daily_volatility, 2), "%")
```

```python
stock_df["Cumulative_Return"] = (
    (stock_df["Close"] / stock_df["Close"].iloc[0]) - 1
) * 100

print(
    stock_df[["Date", "Close", "Cumulative_Return"]].to_string(index=False)
)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(stock_df["Date"], stock_df["Open"], marker="o", label="Open")
plt.plot(stock_df["Date"], stock_df["High"], marker="o", label="High")
plt.plot(stock_df["Date"], stock_df["Low"], marker="o", label="Low")
plt.plot(stock_df["Date"], stock_df["Close"], marker="o", label="Close")

plt.title("Comparison of Stock Prices")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    stock_df["Date"],
    stock_df["Volume"]
)

plt.title("Daily Trading Volume")
plt.xlabel("Date")
plt.ylabel("Trading Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_day = stock_df.loc[stock_df["Daily_Return"].idxmax()]
worst_day = stock_df.loc[stock_df["Daily_Return"].idxmin()]

print("BEST TRADING DAY")
print("Date:", best_day["Date"].date())
print("Closing Price:", best_day["Close"])
print("Daily Return:", round(best_day["Daily_Return"], 2), "%")

print("\nWORST TRADING DAY")
print("Date:", worst_day["Date"].date())
print("Closing Price:", worst_day["Close"])
print("Daily Return:", round(worst_day["Daily_Return"], 2), "%")
```

```python
print("Missing Values:")
print(stock_df.isnull().sum())

print("\nDuplicate Rows:", stock_df.duplicated().sum())

print("\nDuplicate Dates:", stock_df["Date"].duplicated().sum())

print("\nInvalid Price Checks:")
print("High below Open:", (stock_df["High"] < stock_df["Open"]).sum())
print("High below Close:", (stock_df["High"] < stock_df["Close"]).sum())
print("Low above Open:", (stock_df["Low"] > stock_df["Open"]).sum())
print("Low above Close:", (stock_df["Low"] > stock_df["Close"]).sum())
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=stock_df, x="Close", bins=7, kde=True)

plt.title("Distribution of Stock Closing Prices")
plt.xlabel("Closing Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

correlation = stock_df[
    ["Open", "High", "Low", "Close", "Volume", "Daily_Return"]
].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Stock Market Correlation Heatmap")
plt.tight_layout()
plt.show()
```
