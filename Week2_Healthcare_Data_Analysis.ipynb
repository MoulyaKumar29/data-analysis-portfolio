```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

print("Healthcare analysis libraries imported successfully!")
```

```python
healthcare_data = {
    "Patient_ID": list(range(1001, 1021)),
    "Age": [
        25, 34, 45, 52, 60, 29, 41, 55, 67, 38,
        48, 72, 31, 59, 63, 27, 44, 50, 69, 36
    ],
    "Gender": [
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],
    "Medical_Condition": [
        "Diabetes", "Hypertension", "Asthma", "Diabetes",
        "Heart Disease", "Asthma", "Hypertension", "Diabetes",
        "Heart Disease", "Asthma", "Diabetes", "Hypertension",
        "Asthma", "Heart Disease", "Diabetes", "Hypertension",
        "Asthma", "Diabetes", "Heart Disease", "Hypertension"
    ],
    "Blood_Pressure": [
        120, 145, 118, 150, 160, 115, 140, 155, 165, 122,
        148, 170, 119, 158, 152, 135, 125, 149, 162, 138
    ],
    "Glucose_Level": [
        90, 105, 95, 180, 110, 88, 102, 195, 115, 92,
        175, 108, 97, 120, 185, 100, 94, 190, 112, 103
    ],
    "Treatment": [
        "Medication", "Lifestyle", "Medication", "Medication",
        "Surgery", "Lifestyle", "Medication", "Medication",
        "Surgery", "Lifestyle", "Medication", "Lifestyle",
        "Medication", "Surgery", "Medication", "Lifestyle",
        "Medication", "Medication", "Surgery", "Lifestyle"
    ],
    "Recovery_Status": [
        "Recovered", "Improved", "Recovered", "Improved",
        "Improved", "Recovered", "Improved", "Recovered",
        "Improved", "Recovered", "Improved", "Improved",
        "Recovered", "Improved", "Recovered", "Improved",
        "Recovered", "Improved", "Improved", "Recovered"
    ]
}

healthcare_df = pd.DataFrame(healthcare_data)

healthcare_df
```

```python
print("Dataset Shape:", healthcare_df.shape)

print("\nColumn Names:")
print(healthcare_df.columns)

print("\nDataset Information:")
healthcare_df.info()
```

```python
healthcare_df.describe()
```

```python
average_age = healthcare_df["Age"].mean()
average_bp = healthcare_df["Blood_Pressure"].mean()
average_glucose = healthcare_df["Glucose_Level"].mean()

print("Average Patient Age:", average_age)
print("Average Blood Pressure:", average_bp)
print("Average Glucose Level:", average_glucose)
```

```python
condition_counts = healthcare_df["Medical_Condition"].value_counts()

print("Number of Patients by Medical Condition:")
print(condition_counts)
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=healthcare_df,
    x="Medical_Condition",
    order=healthcare_df["Medical_Condition"].value_counts().index
)

plt.title("Number of Patients by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
gender_counts = healthcare_df["Gender"].value_counts()

print("Patient Gender Distribution:")
print(gender_counts)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=healthcare_df, x="Gender")

plt.title("Patient Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(healthcare_df["Age"], bins=6, edgecolor="black")

plt.title("Age Distribution of Patients")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

```python
avg_bp_by_condition = healthcare_df.groupby(
    "Medical_Condition"
)["Blood_Pressure"].mean()

print("Average Blood Pressure by Medical Condition:")
print(avg_bp_by_condition)
```

```python
plt.figure(figsize=(9, 5))

avg_bp_by_condition.sort_values().plot(kind="bar")

plt.title("Average Blood Pressure by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Blood Pressure (mmHg)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
avg_glucose_by_condition = healthcare_df.groupby(
    "Medical_Condition"
)["Glucose_Level"].mean()

print("Average Glucose Level by Medical Condition:")
print(avg_glucose_by_condition)
```

```python
plt.figure(figsize=(9, 5))

avg_glucose_by_condition.sort_values().plot(kind="bar")

plt.title("Average Glucose Level by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Glucose Level")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
recovery_analysis = healthcare_df.groupby(
    "Treatment"
)["Recovery_Status"].value_counts()

print("Recovery Status by Treatment:")
print(recovery_analysis)
```

```python
recovery_table = pd.crosstab(
    healthcare_df["Treatment"],
    healthcare_df["Recovery_Status"]
)

recovery_table.plot(kind="bar", stacked=True, figsize=(8, 5))

plt.title("Recovery Status by Treatment")
plt.xlabel("Treatment")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.legend(title="Recovery Status")
plt.tight_layout()
plt.show()
```

```python
print("Missing Values in Each Column:")
print(healthcare_df.isnull().sum())
```

```python
print("Number of Duplicate Rows:", healthcare_df.duplicated().sum())
```

```python
print("Duplicate Patient IDs:", healthcare_df["Patient_ID"].duplicated().sum())
```

```python
numerical_data = healthcare_df[
    ["Age", "Blood_Pressure", "Glucose_Level"]
]

correlation_matrix = numerical_data.corr()

plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Healthcare Data")
plt.tight_layout()
plt.show()
```

```python
age_bins = [0, 30, 45, 60, 100]
age_labels = ["0–30", "31–45", "46–60", "61–100"]

healthcare_df["Age_Group"] = pd.cut(
    healthcare_df["Age"],
    bins=age_bins,
    labels=age_labels
)

print(healthcare_df[["Age", "Age_Group"]])
```

```python
age_group_counts = healthcare_df["Age_Group"].value_counts().sort_index()

print("Patients by Age Group:")
print(age_group_counts)
```

```python
plt.figure(figsize=(8, 5))

age_group_counts.plot(kind="bar")

plt.title("Patient Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
recovery_table = pd.crosstab(
    healthcare_df["Treatment"],
    healthcare_df["Recovery_Status"]
)

recovery_rate = (
    recovery_table["Recovered"] / recovery_table.sum(axis=1)
) * 100

print("Recovery Rate by Treatment (%):")
print(recovery_rate.round(2))
```

```python
plt.figure(figsize=(8, 5))

recovery_rate.plot(kind="bar")

plt.title("Recorded Recovery Percentage by Treatment")
plt.xlabel("Treatment")
plt.ylabel("Recovered Patients (%)")
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
recovered_glucose = healthcare_df.loc[
    healthcare_df["Recovery_Status"] == "Recovered",
    "Glucose_Level"
]

improved_glucose = healthcare_df.loc[
    healthcare_df["Recovery_Status"] == "Improved",
    "Glucose_Level"
]

t_statistic, p_value = stats.ttest_ind(
    recovered_glucose,
    improved_glucose,
    equal_var=False
)

print("Mean Glucose — Recovered:", recovered_glucose.mean())
print("Mean Glucose — Improved:", improved_glucose.mean())
print("T-statistic:", round(t_statistic, 4))
print("P-value:", round(p_value, 4))
```

```python
total_patients = healthcare_df["Patient_ID"].nunique()
average_age = healthcare_df["Age"].mean()
average_bp = healthcare_df["Blood_Pressure"].mean()
average_glucose = healthcare_df["Glucose_Level"].mean()
recovered_patients = (
    healthcare_df["Recovery_Status"] == "Recovered"
).sum()
improved_patients = (
    healthcare_df["Recovery_Status"] == "Improved"
).sum()

print("HEALTHCARE DATA ANALYSIS SUMMARY")
print("-" * 35)
print("Total Patients:", total_patients)
print(f"Average Age: {average_age:.2f} years")
print(f"Average Blood Pressure: {average_bp:.2f}")
print(f"Average Glucose Level: {average_glucose:.2f}")
print("Recovered Patients:", recovered_patients)
print("Improved Patients:", improved_patients)
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=healthcare_df,
    x="Blood_Pressure",
    y="Glucose_Level",
    hue="Medical_Condition",
    s=100
)

plt.title("Blood Pressure vs Glucose Level")
plt.xlabel("Blood Pressure")
plt.ylabel("Glucose Level")
plt.tight_layout()
plt.show()
```
