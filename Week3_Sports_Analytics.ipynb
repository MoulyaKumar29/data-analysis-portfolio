```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sports_data = {
    "Player": [
        "Virat Kohli", "Rohit Sharma", "KL Rahul", "Hardik Pandya",
        "Jasprit Bumrah", "Shubman Gill", "Suryakumar Yadav",
        "Ravindra Jadeja", "Rishabh Pant", "Mohammed Siraj",
        "MS Dhoni", "Yuvraj Singh", "Shikhar Dhawan",
        "Suresh Raina", "Bhuvneshwar Kumar"
    ],
    "Team": [
        "India", "India", "India", "India", "India",
        "India", "India", "India", "India", "India",
        "India", "India", "India", "India", "India"
    ],
    "Matches": [10, 9, 8, 7, 10, 8, 9, 7, 6, 8, 5, 6, 7, 5, 6],
    "Runs": [
        450, 380, 290, 210, 25, 350, 320, 180,
        240, 15, 150, 190, 280, 170, 20
    ],
    "Wickets": [0, 0, 0, 5, 18, 0, 0, 8, 0, 14, 0, 4, 0, 2, 12],
    "Strike_Rate": [
        140.5, 135.2, 128.4, 155.6, 85.0,
        132.8, 180.2, 115.5, 145.3, 75.0,
        125.0, 130.4, 138.6, 142.1, 80.5
    ]
}

sports_df = pd.DataFrame(sports_data)

sports_df.head()
```

```python
print("Dataset Shape:", sports_df.shape)

print("\nColumn Names:")
print(sports_df.columns)

print("\nDataset Information:")
sports_df.info()

print("\nStatistical Summary:")
sports_df.describe()
```

```python
print("Top 5 Players by Runs:")
print(sports_df.nlargest(5, "Runs")[["Player", "Runs"]])

print("\nTop 5 Players by Wickets:")
print(sports_df.nlargest(5, "Wickets")[["Player", "Wickets"]])

print("\nAverage Runs:", sports_df["Runs"].mean())

print("Average Wickets:", sports_df["Wickets"].mean())

print("Highest Runs:", sports_df["Runs"].max())

print("Highest Wickets:", sports_df["Wickets"].max())
```

```python
top_players = sports_df.nlargest(5, "Runs")

plt.figure(figsize=(10, 6))
sns.barplot(data=top_players, x="Player", y="Runs")

plt.title("Top 5 Players by Runs")
plt.xlabel("Player")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
top_wicket_takers = sports_df.nlargest(5, "Wickets")

plt.figure(figsize=(10, 6))
sns.barplot(data=top_wicket_takers, x="Player", y="Wickets")

plt.title("Top 5 Players by Wickets")
plt.xlabel("Player")
plt.ylabel("Total Wickets")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sports_df["Runs_Per_Match"] = sports_df["Runs"] / sports_df["Matches"]

print("Player Performance:")
print(
    sports_df[["Player", "Matches", "Runs", "Runs_Per_Match"]]
    .sort_values(by="Runs_Per_Match", ascending=False)
    .round(2)
)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sports_df,
    x="Strike_Rate",
    y="Runs",
    hue="Player",
    s=100
)

plt.title("Player Performance: Runs vs Strike Rate")
plt.xlabel("Strike Rate")
plt.ylabel("Total Runs")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=sports_df,
    x="Player",
    y="Runs_Per_Match"
)

plt.title("Average Runs Per Match by Player")
plt.xlabel("Player")
plt.ylabel("Runs Per Match")
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
team_performance = sports_df.groupby("Team").agg({
    "Matches": "sum",
    "Runs": "sum",
    "Wickets": "sum",
    "Strike_Rate": "mean"
}).round(2)

print("Team Performance Summary:")
print(team_performance)
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sports_df,
    x="Runs",
    y="Wickets",
    s=120
)

plt.title("Comparison of Runs and Wickets")
plt.xlabel("Total Runs")
plt.ylabel("Total Wickets")
plt.tight_layout()
plt.show()
```

```python
high_strike_rate = sports_df.sort_values(
    by="Strike_Rate",
    ascending=False
)

print("Players Ranked by Strike Rate:")
print(
    high_strike_rate[
        ["Player", "Runs", "Strike_Rate"]
    ].to_string(index=False)
)
```

```python
print("Missing Values:")
print(sports_df.isnull().sum())

print("\nDuplicate Rows:", sports_df.duplicated().sum())

print("Duplicate Player Names:", sports_df["Player"].duplicated().sum())
```
