The objective of this portfolio is to demonstrate practical skills in Python programming, data manipulation, statistical analysis, data visualization, and business insight generation.

## Project 1: Business Sales Analysis

**Domain:** Business Analytics

**Objective:** Analyze sales data to identify revenue trends, top-performing products, category performance, and regional sales.

**Methodology:**
- Loaded and explored the sales dataset using Pandas.
- Checked data quality and summarized sales information.
- Analyzed revenue by product, category, and region.
- Created visualizations to understand sales performance.

**Business Insights:**
- Identified products and categories contributing to revenue.
- Compared regional sales performance.
- Used visual analysis to understand business sales patterns.

## Project 2: Healthcare Data Analysis

**Domain:** Healthcare Analytics

**Objective:** Explore healthcare data to understand patient characteristics, medical measurements, and treatment outcomes.

**Methodology:**
- Loaded and cleaned sample healthcare data.
- Explored patient characteristics and medical measurements.
- Compared treatment outcomes and recovery.
- Used descriptive statistics and a statistical test.

**Insights:**
- Examined patterns in patient conditions and gender.
- Compared blood pressure and glucose measurements.
- Explored differences in treatment outcomes.

## Project 3: Sports Analytics

**Domain:** Sports Analytics

**Objective:** Analyze cricket player statistics to compare individual performance.

- Compared player performance using charts.

**Insights:**
- Identified high-scoring players.
- Compared bowling performance.
- Explored differences in player statistics.

**Methodology:**

**Insights:**

### General Analysis Workflow

1. Import the required Python libraries.
2. Load datasets using Pandas.
3. Explore the dataset structure and data types.
4. Check for missing values and data quality.
5. Perform data manipulation and analysis.
6. Calculate relevant statistics and metrics.
7. Create charts and visualizations.
8. Interpret the results and document insights.

### Visualization Guidelines

- Use bar charts to compare categories.
- Use line charts to display trends over time.
- Use histograms to understand distributions.
- Use scatter plots to explore relationships.
- Use heatmaps to examine correlations.
- Include meaningful chart titles, axis labels, and legends.

### Data Source

All five projects use synthetic sample datasets created for learning and demonstration. They do not represent real customer, patient, sports, or financial records.

## Setup and Installation

### Prerequisites

- Python 3.10 or later
- Visual Studio Code
- Git
- Jupyter Notebook

### Step 1: Clone the Repository

```bash
git clone https://github.com/MoulyaKumar29/data-analysis-portfolio.git

### Step 2: Open the Project Folder

```bash
cd data-analysis-portfolio

### Step 3: Create a Virtual Environment

Open the VS Code terminal inside the project folder and run:

```powershell
python -m venv .venv
```

### Step 4: Activate the Virtual Environment

For Windows PowerShell, run the following command in the VS Code terminal:

```powershell
.venv\Scripts\Activate.ps1
```

After activation, the terminal should display `(.venv)` before the folder path.

This ensures that the project uses its own Python environment and installed dependencies.

### Step 5: Install Dependencies

After activating the virtual environment, install the required Python libraries using:

```powershell
pip install -r requirements.txt
```

The `requirements.txt` file contains the libraries needed for the portfolio projects.

### Step 6: Launch Jupyter Notebook

Start Jupyter Notebook using:

```powershell
jupyter notebook
```

Open the required project notebook from its respective project folder.

### Step 7: Select the Python Kernel

In VS Code, select the Python interpreter from the `.venv` environment to run the notebooks.
